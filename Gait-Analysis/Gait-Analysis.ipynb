```python
import numpy as np
import pandas as pd
from pathlib import Path

root = Path("C:\\Users\\Lucy\\Desktop\\Uni App\\Gait-Analysis\\UCI HAR Dataset")
```

```python
def loadSignals(split):
    names = [f"{s}_{a}_{split}.txt"
             for s in ("total_acc", "body_acc", "body_gyro")
             for a in ("x", "y", "z")]
    arrays = [np.loadtxt(root / split / "Inertial Signals" / n) for n in names]
    return np.stack(arrays, axis=-1)          # (n_windows, 128, 9)

def loadSplit(split):
    X = loadSignals(split)
    y = np.loadtxt(root / split / f"y_{split}.txt", dtype=int) - 1   # 0..5
    subj = np.loadtxt(root / split / f"subject_{split}.txt", dtype=int)
    return X, y, subj

X_train, y_train, s_train = loadSplit("train")
X_test,  y_test,  s_test  = loadSplit("test")

labels = pd.read_csv(root / "activity_labels.txt", sep=r"\s+", header=None)[1].tolist()
print(X_train.shape, X_test.shape, labels)
```
